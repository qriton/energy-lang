## Line magic — single commands

`auto` as the strength picks the smallest strength that does the edit. Every write is gated: if it would disturb the existing basins it prints `REFUSED` and the layer stays unchanged.

```python
%hlm survey 0
```

```python
%hlm inject 0 42 auto
```

## Cell magic — multi-command scripts inline

```python
%%hlm
restore 0
capture 5 polite Thank you so much for your help
capture 5 polite I really appreciate your patience
capture 5 polite That is very kind of you
inject-concept 5 polite auto
apply 5
generate Tell me about the weather
```

## Write a .hlm script file

```python
%%writefile make_polite.hlm
# Make a model more polite
load model.pt

# Capture examples
capture 5 polite Thank you so much
capture 5 polite I really appreciate it
capture 5 polite That's very kind of you

# Inject the concept (auto strength; REFUSED and unchanged if it would
# disturb the existing basins)
inject-concept 5 polite auto
apply 5

# Verify
benchmark
generate Tell me about the weather

# Cleanup
restore 5
```

```python
# Run it from CLI:
# !qriton-hlm --script make_polite.hlm

# Or from Python:
from qriton_hlm.cli import EnergyLang
lang = EnergyLang()
lang.run_script('make_polite.hlm')
```
